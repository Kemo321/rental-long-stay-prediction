# Generate traffic

```python
import requests
import random
import numpy as np

URL = "http://localhost:8080/predict"

print("Starting 'production' traffic simulation (500 requests)...")

for i in range(500):
    # Instead of uniform, use numpy for skewed distributions
    # Business clients book much earlier (e.g., ~6 months ahead)
    lead_time = int(np.random.exponential(scale=60)) if random.random() > 0.3 else random.randint(100, 360)
    
    # Long stays often correspond to cheaper apartments or, conversely, super-luxurious expensive ones
    price = np.random.lognormal(mean=4.5, sigma=0.8)  # 'Realistic' prices
    
    payload = {
        "accommodates": random.randint(1, 10),  # Expanded domain
        "bathrooms": random.choice([1.0, 1.5, 2.0, 3.0, 4.0]),
        "bedrooms": random.randint(1, 6),
        "beds": random.randint(1, 10),
        "price_num": max(20.0, price),  # Floor at $20
        "number_of_reviews": int(np.random.exponential(scale=50)),  # Many listings have few reviews
        "review_scores_rating": random.choices([0.0, 4.5, 4.8, 5.0], weights=[0.1, 0.2, 0.4, 0.3])[0],
        "availability_rate": random.uniform(0.0, 1.0),
        "comment_length": int(np.random.exponential(scale=100)),
        "has_city_info": random.choice([0, 1]),
        "has_kitchen": random.choices([0, 1], weights=[0.2, 0.8])[0],  # Kitchen is standard
        "has_washer": random.choices([0, 1], weights=[0.4, 0.6])[0],
        "has_workspace": random.choices([0, 1], weights=[0.5, 0.5])[0],
        "amenities_count": random.randint(5, 80),  # Luxury listings have many amenities
        "is_superhost": random.choices([0, 1], weights=[0.7, 0.3])[0],
        "lead_time_days": lead_time,
        "reviews_per_month": random.uniform(0.1, 8.0),
        "calculated_host_listings_count": random.choices([1, 2, 5, 20], weights=[0.6, 0.2, 0.1, 0.1])[0],
        # Intentionally add room type categories (affects the model)
        "room_type_Hotel_room": random.choices([False, True], weights=[0.9, 0.1])[0],
        "room_type_Private_room": random.choices([False, True], weights=[0.7, 0.3])[0],
        "room_type_Shared_room": random.choices([False, True], weights=[0.95, 0.05])[0]
    }

    try:
        response = requests.post(URL, json=payload)
        if response.status_code == 200:
            data = response.json()
            if i % 50 == 0:
                print(f"Progress: {i}/500 requests...")
    except Exception as e:
        print(f"Connection error: {e}")
        break

print("\nDone! New, diverse logs have been generated.")
```
